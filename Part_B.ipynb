1. Compare Normal Distribution vs Standard Normal Distribution:
○ Generate both distributions
○ Plot and explain differences
2. Perform hypothesis testing on two groups:
○ Compute difference in means
○ Interpret results (basic comparison, no advanced formulas required)
3. Explain:
○ When should you standardize data?
○ Why Z-score is important in machine learning?

```python
import random
import matplotlib.pyplot as plt

# Generate Normal Distribution (mean=50, std=10)
data = [random.gauss(50, 10) for _ in range(1000)]

# Compute mean and std manually
mean = sum(data) / len(data)
std = (sum((x - mean)**2 for x in data) / len(data)) ** 0.5

# Convert to Standard Normal (Z-score)
z_data = []
for x in data:
    z = (x - mean) / std
    z_data.append(z)

# Plot both
plt.figure()

plt.hist(data, bins=30, alpha=0.5)
plt.hist(z_data, bins=30, alpha=0.5)

plt.title("Normal vs Standard Normal Distribution")
plt.show()
```

```python
# Two sample groups
group1 = [45, 50, 55, 60, 65]
group2 = [70, 75, 80, 85, 90]

# Compute means
mean1 = sum(group1) / len(group1)
mean2 = sum(group2) / len(group2)

# Difference
difference = mean1 - mean2

print("Mean Group 1:", mean1)
print("Mean Group 2:", mean2)
print("Difference:", difference)
```

Interpretation
If difference ≈ 0 → groups are similar
If difference is large → groups are different

Group 2 has significantly higher mean
So we can conclude: Groups are different

When to Standardize?
You should standardize when:
Features have different scales

Example: Age (0–100) vs Salary (0–1,00,000)

Using distance-based models
KNN
